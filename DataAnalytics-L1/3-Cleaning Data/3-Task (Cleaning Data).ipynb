## 13. Missing Data Handling Strategy
This is an important Markdown cell for your notebook.
Missing Data Strategy
Age
The Age column contains 86 missing values. Age is a numerical variable, and passenger ages are not normally distributed perfectly. Therefore, median imputation is preferred because the median is less affected by extreme ages than the mean.
Fare
Only one value is missing in Fare. Since Fare is numerical, the median Fare is used. This avoids deleting an otherwise useful passenger record.
Cabin
The Cabin column has a large number of missing values. Deleting 327 rows would remove most of the dataset. Therefore, missing cabin information is replaced with "Unknown" rather than deleting rows.
Other columns
Other columns contain no missing values, so no imputation is required.

```python
# Create a copy of the original dataset
cleaned_df = df.copy()

# Check that the copy was created
cleaned_df.head()
```

```python
# Calculate median age
age_median = cleaned_df["Age"].median()

print("Median Age:", age_median)
```

```python
# Fill missing Age values with the median
cleaned_df["Age"] = cleaned_df["Age"].fillna(age_median)

# Check missing values again
print("Missing Age after imputation:", cleaned_df["Age"].isnull().sum())
```

```python
# Calculate median Fare
fare_median = cleaned_df["Fare"].median()

print("Median Fare:", fare_median)
```

```python
# Fill missing Fare values with median Fare
cleaned_df["Fare"] = cleaned_df["Fare"].fillna(fare_median)

# Check missing Fare values
print("Missing Fare after imputation:", cleaned_df["Fare"].isnull().sum())
```

```python
# Replace missing Cabin values with "Unknown"
cleaned_df["Cabin"] = cleaned_df["Cabin"].fillna("Unknown")

# Check missing Cabin values
print("Missing Cabin after cleaning:", cleaned_df["Cabin"].isnull().sum())
```

## Outlier Detection Using IQR
The Interquartile Range (IQR) method is used to identify potential outliers.

```python
# Select numerical columns
numeric_columns = cleaned_df.select_dtypes(include=np.number).columns

# Create an empty dictionary for outlier information
outlier_report = {}

# Loop through numerical columns
for column in numeric_columns:
    
    # Calculate Q1 and Q3
    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)
    
    # Calculate IQR
    IQR = Q3 - Q1
    
    # Calculate lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    # Identify outliers
    outliers = cleaned_df[
        (cleaned_df[column] < lower_limit) |
        (cleaned_df[column] > upper_limit)
    ]
    
    # Store results
    outlier_report[column] = {
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    }

# Convert dictionary into DataFrame
outlier_report_df = pd.DataFrame(outlier_report).T

outlier_report_df
```

# Display number of potential outliers in each numerical column

```python
# Display number of potential outliers in each numerical column

for column in numeric_columns:
    
    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    count = (
        (cleaned_df[column] < lower_limit) |
        (cleaned_df[column] > upper_limit)
    ).sum()
    
    print(column, ":", count, "potential outliers")
```

# Function to check whether a column has the expected dtype

```python
# Function to check whether a column has the expected dtype

def check_dtype_accuracy(dataframe):
    
    results = {}
    
    results["PassengerId"] = str(dataframe["PassengerId"].dtype) == "string"
    results["Survived"] = str(dataframe["Survived"].dtype) == "int64"
    results["Pclass"] = str(dataframe["Pclass"].dtype) == "int64"
    results["Name"] = str(dataframe["Name"].dtype) == "string"
    results["Sex"] = str(dataframe["Sex"].dtype) == "string"
    results["Age"] = str(dataframe["Age"].dtype) == "float64"
    results["SibSp"] = str(dataframe["SibSp"].dtype) == "int64"
    results["Parch"] = str(dataframe["Parch"].dtype) == "int64"
    results["Ticket"] = str(dataframe["Ticket"].dtype) == "string"
    results["Fare"] = str(dataframe["Fare"].dtype) == "float64"
    results["Cabin"] = str(dataframe["Cabin"].dtype) == "string"
    results["Embarked"] = str(dataframe["Embarked"].dtype) == "string"
    
    return results
```
